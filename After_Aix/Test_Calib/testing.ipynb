```python
import uhd

CHANNEL = 0

usrp = uhd.usrp.MultiUSRP()

print("Appareil :")
print(usrp.get_pp_string())

supported = usrp.has_rx_power_reference(CHANNEL)

print("Support RX power reference :", supported)

if supported:
    try:
        power_ref = usrp.get_rx_power_reference(CHANNEL)
        print("Référence RX actuelle :", power_ref, "dBm")
    except RuntimeError as error:
        print("Erreur de calibration :", error)
else:
    print("La référence de puissance RX n'est pas supportée.")
```
